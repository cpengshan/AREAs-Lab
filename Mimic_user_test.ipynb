```python
import json
import ast
from mimic_user_utils import MimicUser
from mimic_user_utils import conduct_interaction


# Load persona from users.json (taking the first one for example)
with open('/content/users.json', 'r') as f:
    users_data = json.load(f)
persona_to_use = users_data[0] # Using the first persona as an example

# Load persona from users.json (taking the first one for example)
with open('/content/responser_habit.json', 'r') as f:
    commnication = json.load(f)
communication_style = commnication[0] # Using the first persona as an example


with open('/content/persona_1.json', 'r') as f:
      raw_content = f.read()
      persona_1_data = ast.literal_eval(raw_content)
      print("Successfully loaded persona_1.json using ast.literal_eval.")


# Assign task_requirement_gold from persona_1.json
task_requirement_gold = persona_1_data.get('task_requirement')

# Assign user_instruction_init from persona_1.json using 'elevator_pitch_summary'
user_instruction_init = persona_1_data.get('elevator_pitch_summary')

communication_style_init = communication_style.get('communication_style')

# Initialize mimic user with the loaded data and template path
mimic_user = MimicUser(
    persona=persona_to_use,
    task_requirement_gold=task_requirement_gold,
    user_instruction_init=user_instruction_init,
    communication_style_init = communication_style_init,
    model="gpt-4.1-mini", # Keep the model from previous example
    template_path="/content/feedback.jinja" # Specify the template file
)

#print("MimicUser initialized successfully.")
#print("\nSelected Persona:")
#print(json.dumps(persona_to_use, indent=2))
#print("\nSelected Gold Task Requirement:")
#print(json.dumps(task_requirement_gold, indent=2))
#print("\nSelected User Instruction Init:")
#print(json.dumps(user_instruction_init, indent=2))
#print("\ncommunication Init:")
#print(json.dumps(communication_style_init, indent=2))



```

```python
#from conduct_interaction import conduct_interaction


state = {
    "messages": [
        {
            "role": "agent",
            "content": "What are you trying to achieve overall？" ,
        }
    ],
    "is_complete": False
}

# Refactor using conduct_interaction for turn 1
agent_question_1 = "What are you trying to achieve overall?"
state = conduct_interaction(mimic_user, state, agent_question_1, 1)


# Refactor using conduct_interaction for turn 2
if not state["is_complete"]:
    agent_question_2 = "What does your input data look like?"
    state = conduct_interaction(mimic_user, state, agent_question_2, 2)

# Third round
agent_question_3 = "What format should the output be in?"
state = conduct_interaction(mimic_user, state, agent_question_3, 3)

# Fourth round
if not state["is_complete"]:
    agent_question_4 = "Is it clean, or does it contain noise / multiple sources？"
    state = conduct_interaction(mimic_user, state, agent_question_4, 4)

agent_question_5 = "Are there any important details that must not be lost, even if they appear only once?"
state = conduct_interaction(mimic_user, state, agent_question_5, 5)


agent_question_6 = "How would you judge if the result is good?"
state = conduct_interaction(mimic_user, state, agent_question_6, 6)
```
