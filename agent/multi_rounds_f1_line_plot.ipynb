```python
import json
import matplotlib.pyplot as plt
import numpy as np

base = "/local/scratch/zzh2365/AUNU/agent/results copy/alexfabbri_multi_news/data_multi_rounds"

steps = list(range(1, 6))
exp_nums = list(range(4, 9))  # Experiment4=Step1 ... Experiment8=Step5
case_ids = ['1', '2', '3', '4', '5']

# Average F1 across task_1 and task_2 per case per step
case_f1 = {c: [] for c in case_ids}

for exp_num in exp_nums:
    with open(f"{base}/Experiment{exp_num}/eval_results.json") as f:
        data = json.load(f)
    for c in case_ids:
        f1s = [data[c][t]['scores']['f1'] for t in data[c]]
        case_f1[c].append(np.mean(f1s))

fig, ax = plt.subplots(figsize=(8, 5))
colors = plt.cm.tab10.colors

for i, c in enumerate(case_ids):
    ax.plot(steps, case_f1[c], marker='o', label=f'Case {c}', color=colors[i])

ax.set_xlabel('Step', fontsize=12)
ax.set_ylabel('F1 Score', fontsize=12)
ax.set_title('F1 Score per Case Across Steps\n(alexfabbri_multi_news / data_multi_rounds)', fontsize=11)
ax.set_xticks(steps)
ax.set_xticklabels([f'Step {s}' for s in steps])
ax.legend(title='Case')
ax.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig("/local/scratch/zzh2365/AUNU/agent/multi_rounds_f1_line_plot.png", dpi=150)
plt.show()
print("Saved to agent/multi_rounds_f1_line_plot.png")
```
